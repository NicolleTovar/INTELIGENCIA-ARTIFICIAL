## Resumen: Support Vector Machine (SVM)

 **Support Vector Machine (SVM)** es un algoritmo de **aprendizaje supervisado** utilizado principalmente para tareas de **clasificación** y, en algunos casos, **regresión**. Su objetivo es encontrar una frontera que separe de la mejor manera posible los datos pertenecientes a diferentes clases.

 ### ¿Cómo funciona?

 En un problema de clasificación, SVM busca un **hiperplano** que divida los datos en dos o más grupos. La característica principal es que intenta **maximizar el margen**, es decir, la distancia entre la frontera de decisión y los puntos más cercanos de cada clase.

 Los puntos que se encuentran más cerca de la frontera se denominan **vectores de soporte**, porque son los que determinan principalmente la posición del hiperplano.

 ### Conceptos principales

 - **Hiperplano:** superficie que separa las diferentes clases.
- **Margen:** distancia entre el hiperplano y los ejemplos más cercanos.
- **Vectores de soporte:** datos más cercanos al hiperplano y fundamentales para definirlo.
- **Kernel:** función que permite transformar los datos para encontrar separaciones no lineales.
- **C:** parámetro que controla el equilibrio entre maximizar el margen y clasificar correctamente los datos.
- **Gamma (γ):** en determinados kernels, controla la influencia de cada punto sobre la frontera de decisión.

 ### Kernel

 Una de las ventajas de SVM es que puede trabajar con datos que no son linealmente separables mediante diferentes funciones **kernel**, por ejemplo:

 - **Lineal:** adecuado cuando las clases pueden separarse mediante una línea o hiperplano.
- **Polinomial:** permite fronteras de decisión más complejas.
- **RBF (Radial Basis Function):** muy utilizado para problemas con relaciones no lineales.
- **Sigmoide:** utiliza una función similar a la empleada en redes neuronales.

 ### Ventajas

 - Puede funcionar muy bien con **conjuntos de datos pequeños o medianos**.
- Es efectivo cuando existen muchas **características (features)**.
- Puede construir fronteras de decisión **no lineales** mediante kernels.
- Busca una solución que maximice el margen, lo que puede favorecer la generalización.

 ### Desventajas

 - Puede ser **costoso computacionalmente** con conjuntos de datos muy grandes.
- Su rendimiento depende de una adecuada selección de parámetros como **C y γ**.
- Los resultados pueden ser sensibles a la **escala de las variables**, por lo que normalmente se recomienda estandarizar los datos.
- Los modelos con kernels no lineales pueden ser más difíciles de interpretar.

 ### Aplicaciones

 SVM se ha utilizado en áreas como:

 - Clasificación de textos y detección de spam.
- Reconocimiento de imágenes.
- Clasificación de datos médicos.
- Reconocimiento de patrones.
- Análisis de datos financieros.
- Clasificación de documentos.

 **En resumen**, Support Vector Machine es un algoritmo que busca separar las clases mediante una frontera que maximiza el margen entre ellas. Su capacidad para utilizar kernels le permite resolver tanto problemas lineales como no lineales, convirtiéndolo en una herramienta importante dentro del aprendizaje automático.
