**Árboles de decisión**
Los árboles de decisión son modelos de aprendizaje automático que se utilizan para resolver problemas de clasificación y regresión. Están formados por nodos, ramas y hojas, y toman decisiones mediante condiciones aplicadas a las características de los datos hasta obtener una predicción.

**Modelos de caja blanca**
Se consideran modelos de caja blanca porque su funcionamiento es fácil de interpretar. Se puede observar qué características utiliza el árbol, las condiciones de cada división y el recorrido que sigue cada dato hasta llegar a una predicción.

**Entrenamiento**
Durante el entrenamiento, el algoritmo busca las mejores formas de dividir los datos. En clasificación, intenta separar las diferentes clases de la manera más adecuada, utilizando medidas como Gini y entropía para evaluar la calidad de las divisiones.

El árbol continúa creciendo hasta alcanzar las condiciones de parada establecidas.

Predicciones y probabilidades
Una vez entrenado, el árbol puede realizar predicciones sobre nuevos datos. En clasificación, devuelve una categoría o clase, mientras que en regresión proporciona un valor numérico.

En los árboles de clasificación también es posible obtener probabilidades para cada clase. Estas se calculan según la distribución de las clases de los datos que llegan a la hoja correspondiente.

CART
CART (Classification and Regression Trees) es un método utilizado para construir árboles de decisión tanto de clasificación como de regresión. En clasificación suele utilizar el índice Gini para seleccionar las divisiones, mientras que en regresión busca reducir el error o la variabilidad dentro de los grupos.

Regularización
Si un árbol crece demasiado, puede aprender detalles específicos de los datos de entrenamiento y producir sobreajuste (overfitting). Para evitarlo, se puede controlar su tamaño y complejidad mediante parámetros como:

max_depth: limita la profundidad del árbol.

min_samples_split: establece el mínimo de observaciones necesarias para dividir un nodo.

min_samples_leaf: indica el mínimo de observaciones que puede tener una hoja.

max_leaf_nodes: limita la cantidad de hojas.

ccp_alpha: controla la poda mediante el método de complejidad de costo.

Árboles de regresión
Los árboles de regresión se utilizan cuando la variable objetivo es numérica. El árbol divide los datos en diferentes grupos y cada hoja genera un valor de predicción. Durante el entrenamiento, busca divisiones que reduzcan el error dentro de cada grupo.

Resumen general
Los árboles de decisión son modelos interpretables y versátiles que permiten realizar tareas de clasificación y regresión. Pueden generar predicciones y probabilidades en clasificación y cuentan con técnicas de regularización para reducir el sobreajuste. CART es uno de los métodos principales utilizados para construir estos árboles.
