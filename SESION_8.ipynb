# MNIST: reconocimiento de imágenes y aprendizaje automático

## Introducción

En el desarrollo de la Inteligencia Artificial y el aprendizaje automático, uno de los principales objetivos consiste en crear sistemas capaces de interpretar información de manera similar a como lo hacen las personas. Una de las áreas más importantes para alcanzar este objetivo es el reconocimiento de imágenes.

Los seres humanos pueden observar una imagen y reconocer fácilmente un número escrito a mano. Sin embargo, para un computador esta tarea representa un problema mucho más complejo, debido a que una imagen está formada por una gran cantidad de valores numéricos que deben ser analizados para determinar qué representa.

Para estudiar este tipo de problemas se creó **MNIST**, uno de los conjuntos de datos más conocidos en el campo del aprendizaje automático y el reconocimiento de imágenes. MNIST ha sido utilizado durante muchos años como referencia para desarrollar, probar y comparar diferentes algoritmos de clasificación.

El conjunto de datos contiene imágenes de números escritos a mano, correspondientes a los dígitos del **0 al 9**. Cada imagen puede ser procesada por un algoritmo para intentar determinar qué número representa.

MNIST permite comprender conceptos fundamentales de la Ciencia de Datos, como la preparación de datos, el entrenamiento de modelos, la clasificación, la evaluación y el reconocimiento de patrones.

---

# ¿Qué es MNIST?

MNIST es un conjunto de datos diseñado para trabajar con el reconocimiento automático de dígitos escritos a mano.

El nombre MNIST proviene de **Modified National Institute of Standards and Technology**.

El conjunto contiene imágenes de dígitos escritos manualmente y sus respectivas etiquetas. Cada imagen corresponde a uno de los diez números posibles:

**0, 1, 2, 3, 4, 5, 6, 7, 8 y 9.**

Cada imagen tiene un tamaño de **28 × 28 píxeles**, lo que significa que contiene:

**28 × 28 = 784 píxeles.**

Cada píxel representa un valor numérico relacionado con la intensidad de la imagen.

Por lo tanto, para un modelo de aprendizaje automático, una imagen puede convertirse en un conjunto de **784 valores numéricos** que representan sus píxeles.

Esto permite que un algoritmo matemático pueda trabajar con la imagen.

---

# ¿Por qué MNIST es importante?

MNIST es importante porque proporciona un problema relativamente sencillo para estudiar técnicas de aprendizaje automático.

Antes de trabajar con imágenes mucho más complejas, MNIST permite comprender cómo un modelo puede aprender a reconocer patrones.

Por ejemplo, una persona puede observar una imagen y reconocer inmediatamente que representa el número 7.

Un algoritmo, en cambio, debe aprender cuáles son las características que diferencian un 7 de un 1, un 2, un 9 u otro número.

El modelo debe analizar numerosos ejemplos para identificar patrones.

Por esta razón, MNIST es utilizado frecuentemente como un punto de partida para estudiar:

* Machine Learning.
* Deep Learning.
* Redes neuronales.
* Clasificación.
* Reconocimiento de imágenes.
* Procesamiento de imágenes.
* Inteligencia Artificial.

# Características del conjunto de datos MNIST

MNIST contiene **70.000 imágenes** de dígitos escritos a mano.

Tradicionalmente, el conjunto se divide en:

* **60.000 imágenes para entrenamiento.**
* **10.000 imágenes para prueba.**

Las imágenes están organizadas en diez categorías correspondientes a los números del 0 al 9.

Cada imagen tiene:

**28 píxeles de ancho × 28 píxeles de alto.**

Por lo tanto:

**28 × 28 = 784 características por imagen.**

Cada una de estas características corresponde a un píxel.

El modelo recibe estos valores y aprende a relacionarlos con una etiqueta.

Por ejemplo:

**Imagen → 784 valores → Modelo → Predicción → Dígito**

---

# ¿Cómo representa un computador una imagen MNIST?

Para una persona, una imagen representa visualmente un número.

Para un computador, la imagen puede representarse como una matriz de números.

Una imagen de 28 × 28 píxeles puede representarse de esta manera:

**Fila 1:** valores de los píxeles

**Fila 2:** valores de los píxeles

**Fila 3:** valores de los píxeles

...

**Fila 28:** valores de los píxeles

Cada valor representa la intensidad de un píxel.

En una imagen en escala de grises, los valores pueden representar diferentes niveles de intensidad.

De esta manera, una imagen puede convertirse en información numérica que un algoritmo puede procesar.

---

# Las etiquetas de MNIST

Cada imagen del conjunto de datos tiene una etiqueta que indica qué número representa.

Por ejemplo:

**Imagen de un número → etiqueta 5**

Otro ejemplo:

**Imagen de un número → etiqueta 8**

Estas etiquetas son fundamentales durante el entrenamiento porque permiten que el modelo compare sus predicciones con la respuesta correcta.

El proceso puede representarse así:

**Imagen + etiqueta correcta → entrenamiento → aprendizaje del patrón**

Posteriormente:

**Nueva imagen → modelo → predicción**

---

# MNIST como problema de clasificación

MNIST es principalmente un problema de **clasificación supervisada**.

La clasificación consiste en asignar una entrada a una categoría determinada.

En este caso existen diez categorías:

**0, 1, 2, 3, 4, 5, 6, 7, 8 y 9.**

El modelo recibe una imagen y debe determinar a cuál de las diez clases pertenece.

Por ejemplo:

**Entrada:** imagen de un dígito.

**Salida esperada:** 4.

Si el modelo identifica correctamente la imagen, la clasificación es correcta.

Si determina que es otro número, la predicción es incorrecta.

---

# Aprendizaje supervisado en MNIST

MNIST es un ejemplo clásico de aprendizaje supervisado.

En el aprendizaje supervisado, el modelo recibe ejemplos que contienen tanto los datos como las respuestas correctas.

En MNIST:

**Datos:** imágenes de dígitos.

**Etiquetas:** número correspondiente a cada imagen.

El modelo analiza muchos ejemplos y busca aprender la relación entre las imágenes y sus etiquetas.

Por ejemplo:

**Imagen → 2**

**Imagen → 7**

**Imagen → 3**

**Imagen → 9**

Después de analizar numerosos ejemplos, el modelo intenta aprender las características que permiten diferenciar los diferentes números.

---

# Etapa de entrenamiento

Durante el entrenamiento, el modelo recibe las imágenes y las etiquetas correspondientes.

El objetivo es que aprenda patrones que le permitan realizar correctamente la clasificación.

Un proceso simplificado sería:

1. Se proporciona una imagen.
2. El modelo analiza sus píxeles.
3. El modelo genera una predicción.
4. La predicción se compara con la etiqueta correcta.
5. Se calcula el error.
6. El modelo ajusta sus parámetros.
7. El proceso se repite con muchos ejemplos.

Después de repetir este proceso muchas veces, el modelo puede mejorar su capacidad para reconocer los dígitos.

---

# Etapa de prueba

Una vez entrenado el modelo, es necesario comprobar qué tan bien funciona con imágenes que no utilizó durante el entrenamiento.

Para esto se utiliza el conjunto de prueba.

La idea es evaluar si el modelo realmente aprendió patrones generales o simplemente memorizó los ejemplos de entrenamiento.

Por ejemplo, si el modelo obtiene buenos resultados con las imágenes de entrenamiento pero resultados deficientes con imágenes nuevas, puede existir un problema de **sobreajuste**.

El conjunto de prueba permite medir el comportamiento del modelo frente a datos que no ha visto previamente.

---

# ¿Qué es el sobreajuste?

El sobreajuste, conocido como **overfitting**, ocurre cuando un modelo aprende demasiado específicamente los datos de entrenamiento.

En lugar de aprender patrones generales, puede aprender características particulares de los ejemplos utilizados.

Esto puede provocar que el modelo tenga:

**Excelente desempeño en entrenamiento**

pero

**Bajo desempeño en datos nuevos.**

En un proyecto de reconocimiento de imágenes, esto representa un problema porque el objetivo es que el modelo pueda reconocer nuevos dígitos, no solamente repetir los ejemplos que ya conoce.

MNIST permite estudiar este fenómeno y comprender la importancia de evaluar los modelos con datos independientes.

# Modelos que pueden utilizarse con MNIST

MNIST puede utilizarse para estudiar diferentes algoritmos de aprendizaje automático.

Entre ellos se encuentran:

* Regresión logística.
* Árboles de decisión.
* K-Nearest Neighbors.
* Máquinas de vectores de soporte.

Cada modelo utiliza estrategias diferentes para encontrar patrones.

---

# Regresión logística

Aunque su nombre contiene la palabra regresión, la regresión logística puede utilizarse para problemas de clasificación.

En MNIST puede utilizarse para determinar a cuál de las diez categorías pertenece una imagen.

El modelo analiza los valores de los píxeles y calcula probabilidades relacionadas con las diferentes clases.

Finalmente, selecciona la categoría con mayor probabilidad.

---

# Árboles de decisión

Los árboles de decisión utilizan una estructura basada en preguntas o condiciones para realizar una clasificación.

En un problema de imágenes, pueden analizar diferentes características de los datos y dividir progresivamente las muestras.

Aunque pueden utilizarse con MNIST, los árboles de decisión no suelen ser la herramienta principal para obtener los mejores resultados en reconocimiento de imágenes.

Sin embargo, son útiles para comprender cómo diferentes algoritmos pueden abordar un mismo problema.

---

# K-Nearest Neighbors

El algoritmo **K-Nearest Neighbors (KNN)** clasifica una nueva imagen comparándola con ejemplos conocidos.

La idea general consiste en buscar ejemplos similares y determinar qué categoría predomina entre ellos.

Por ejemplo, si una nueva imagen es muy similar a varias imágenes etiquetadas como 3, el algoritmo puede clasificarla como un 3.

Este método permite comprender el concepto de similitud entre datos.

---

# Redes neuronales

Las redes neuronales son modelos especialmente importantes para el reconocimiento de imágenes.

Están inspiradas de manera simplificada en el funcionamiento de las redes neuronales biológicas y están formadas por diferentes capas de unidades computacionales.

Una red neuronal puede recibir los valores correspondientes a los píxeles y transformarlos progresivamente para identificar patrones.

Una estructura sencilla podría ser:

**Entrada → Capa oculta → Capa oculta → Salida**

La capa de entrada recibe los datos de la imagen.

Las capas intermedias procesan la información.

La capa de salida genera una predicción entre las diez clases posibles.

---

# Redes neuronales convolucionales

Las **Redes Neuronales Convolucionales**, conocidas como **CNN**, son especialmente importantes en el procesamiento y reconocimiento de imágenes.

Estas redes pueden aprender características visuales de las imágenes.

En las primeras capas pueden identificar características sencillas, como bordes o líneas.

En capas posteriores pueden combinar estas características para reconocer estructuras más complejas.

En MNIST, una CNN puede aprender diferentes características de los dígitos y utilizar esa información para clasificarlos.

Por esta razón, las redes convolucionales son una herramienta muy importante dentro del reconocimiento automático de imágenes.

---

# Evaluación del modelo

Después del entrenamiento, es necesario evaluar el desempeño del modelo.

Una de las métricas más conocidas es la **exactitud o accuracy**.

La exactitud representa la proporción de predicciones correctas respecto al total de predicciones.

De manera simplificada:

**Accuracy = predicciones correctas / total de predicciones**

Por ejemplo, si un modelo clasifica correctamente 9.500 imágenes de un conjunto de 10.000:

**Accuracy = 9.500 / 10.000 = 0,95**

Esto equivale a:

**95 % de exactitud.**

Sin embargo, la exactitud no siempre es suficiente. Dependiendo del problema, también pueden utilizarse otras métricas.

---

# Matriz de confusión

La matriz de confusión es una herramienta que permite analizar con mayor detalle las predicciones de un modelo de clasificación.

En MNIST permite observar cuáles números se confunden con otros.

Por ejemplo, un modelo podría tener mayor dificultad para diferenciar determinadas formas de números escritos a mano.

La matriz de confusión permite identificar:

* Predicciones correctas.
* Errores.
* Clases con mayor dificultad.
* Clases que suelen confundirse entre sí.

Esto ayuda a comprender mejor el comportamiento del modelo.

# Limitaciones de MNIST

A pesar de su importancia educativa, MNIST presenta ciertas limitaciones.

Las imágenes son relativamente sencillas y están estandarizadas.

Los dígitos se encuentran centrados y utilizan una escala de grises.

Por esta razón, un modelo que funciona muy bien en MNIST no necesariamente tendrá el mismo desempeño en imágenes reales mucho más complejas.

En el mundo real pueden existir:

* Diferentes tamaños.
* Diferentes fondos.
* Ruido.
* Iluminación variable.
* Rotaciones.
* Diferentes estilos de escritura.
* Imágenes de mayor resolución.

Por esta razón, MNIST debe considerarse principalmente como un conjunto de datos de referencia y aprendizaje.
